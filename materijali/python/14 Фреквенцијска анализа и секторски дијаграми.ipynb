### 14.1 Фреквенцијска анализа

Фреквенција или учесталост представља број појављивања неког елемента у неком низу. Када имамо мале низове, није проблем да ручно рачунамо феквенције, али када имамо велике сетове података, мали програм који то ради уместо нас нам доста олакшава посао. Посматрајмо сада пример где играч баца коцкицу и постоји 6 могућих исхода тог бацања.

```python
бројеви=[6,5,4,5,4,5,6,3,2,3,1,2,3,4,3,4,5,3,3,3,5,5,5,6,3,4,2,1,1,2,1,6,5,6,4,3,2,3,4,2,1,2,3,4,5,3,2,1,6,4,5,3,5,6,5,6,3,4,2]
```

Сада треба да пребројимо колико се пута који број појавио.

```python
број1=бројеви.count(1)
број2=бројеви.count(2)
број3=бројеви.count(3)
број4=бројеви.count(4)
број5=бројеви.count(5)
број6=бројеви.count(6)
print("Бројеви су распоређени на следећи начин:")
print("Број 1 смо добили",број1,"пута.")
print("Број 2 смо добили",број2,"пута.")
print("Број 3 смо добили",број3,"пута.")
print("Број 4 смо добили",број4,"пута.")
print("Број 5 смо добили",број5,"пута.")
print("Број 6 смо добили",број6,"пута.")
```

Сада можемо и да видимо колики је процентуални удео ког броја међу бацањима.

```python
укупно=len(бројеви)
удео1=100.0*број1/укупно
удео2=100.0*број2/укупно
удео3=100.0*број3/укупно
удео4=100.0*број4/укупно
удео5=100.0*број5/укупно
удео6=100.0*број6/укупно
print("Бројеви су процентуално распоређени на следећи начин:")
print("Број 1 се појављује у ",round(удео1,2),"% случајева.")
print("Број 2 се појављује у ",round(удео2,2),"% случајева.")
print("Број 3 се појављује у ",round(удео3,2),"% случајева.")
print("Број 4 се појављује у ",round(удео4,2),"% случајева.")
print("Број 5 се појављује у ",round(удео5,2),"% случајева.")
print("Број 6 се појављује у ",round(удео6,2),"% случајева.")
```

Као што видимо, помоћу фреквенцијске анализе можемо видети колико често се који број појављује, али можемо уз малу модификацију да прикажемо и релативни удео изражен у процентима.

### 14.2 Секторски дијаграми

Секторски дијаграми или још познати као пита дијаграми су идеални за приказивање удела. Сада ћемо посматрати станаре једне зграде подељене у старосне категорије.

```python
import matplotlib.pyplot as plt
узраст=["0-17","18-30","31-50","51-64","65+"]
број_људи=[20,12,22,18,10]
```

Сада ћемо ове податке да представимо помоћу секторског дијаграма:

```python
plt.figure(figsize=(6,6))
plt.pie(број_људи, labels=узраст)
plt.title("Узраст станара зграде")
plt.show()
plt.close()
```

У случају да неке податке желимо посебно да нагласимо, можемо изместити одрећене делове дијаграма из центра на следећи начин:

```python
plt.figure(figsize=(6,6))
измештање=[0.2,0.2,0,0,0]
plt.pie(број_људи, labels=узраст,explode=измештање)
plt.title("Узраст станара зграде")
plt.show()
plt.close()
```

Посматрајмо сада пример једне фирме и удела у власништву њених власника.

```python
власници=["Самсунг","Тошиба","Моторола","Џон Смит","Пол Адамс","Крис Џонсон"]
удео=[65,18,12,4,0.9,0.1]
```

Хајде сада да представимо секторским дијаграмом њихов удео у власништву:

```python
plt.figure(figsize=(6,6))
plt.pie(удео, labels=власници)
plt.title("Удео у власништву")
plt.show()
plt.close()
```

Може се приметити да се подаци тројице власника са најмањим уделом не виде добро. Зато ћемо тај део дијаграма додатно истаћи.

```python
plt.figure(figsize=(6,6))
измештање=[0,0,0,0,1,1]
plt.pie(удео, labels=власници,explode=измештање)
plt.title("Удео у власништву")
plt.show()
plt.close()
```

#### Задатак 5
