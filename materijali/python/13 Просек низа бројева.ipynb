# 13 Просек низа бројева

### 13.1 Рачунање просека низа

Када имамо неки низ бројева, поред његове дужине, врло често нам је битно да знамо и просек низа. То може бити просек оцена у школи, просек плата у некој фирми или било који други просек који нам даје бољу слику о неком низу бројева. Пошто у Пајтону не постоји функција за рачунање просека као у Екселу, мораћемо сами да је направимо. Познато је да се просек елемената једног низа може израчунати као количник суме елемената тог низа и његовог броја елемената. Показаћемо прво како раде те две функције.

```python
низ=[4,3,5,2,7,10,3]
print(len(низ))
print(sum(низ))
```

Сада једноставно можемо израчунати просек низа.

```python
sum(низ)/len(низ)
```

Постоји начин да ово све изгледа лепше. То је могуће ако дефинишемо функцију која рачуна просек.

```python
def avg(x):
    return sum(x)/len(x)
```

Сада можемо да испишемо просек низа једноставним позивом функције коју смо дефинисали.

```python
avg(низ)
```

Када имамо пуно децимала то делује конфузно и непрегледно, па ћемо овај резултат заокружити на две децимале.

```python
round(avg(низ),2)
```

### 13.2 Сортирање низа и медијана

У ситуацијама када имамо велики број података, да бисмо их боље разумели може да нам помогне сортирање. Под сортирањем подразумевамо урећивање редоследа, тако да бројеви иду од најмањег до највећег или од највећег до најмањег. Пајтон за то има уграђену функцију. Погледајмо пример:

```python
поени=[41,25,34,76,12,98,21,52]
поени.sort()
поени
```

Можемо приметити да су бројеви сортирани од најмањег до највећег. Ако желимо да бројеви буду поређани од највећег ка најмањем треба да урадимо следеће:

```python
поени.sort(reverse=True)
поени
```

Понекад нам просек неког низа може дати лажну слику о том низу. То се обично дешава када имамо податке који доста одступају од просека. У таквим ситуацијама нам је медијана бољи показатељ низа. Медијана представља елемент на средини сортираног низа. То значи да се половина елемената налази испод а половина изнад медијане. Ово је посебно важно када имамо велике количине података. Хајде сада да дефинишемо функцију која рачуна медијану једног низа:

```python
def medijana(x):
    n=len(x)
    x.sort()
    return x[n//2]
```

Посматрајмо сада следећи низ података:

```python
низ=[1,2,3,2,1,2,3,2,1,2,3,2,1,2,2,3,2,1,1,2,2,3,2,1,2,1,2,3,2,1,1,1,2,3,2,1,2,1,2,3,2,1,2,3,2,1,1000,1200]
```

Израчунаћемо сада његов просек и медијану.

```python
avg(низ)
```

```python
medijana(низ)
```

Kao што видимо, просек и медијана се доста разликују. То се десило јер имамо два елемента низа који драстично одступају од осталих вредности у низу.

### 13.3 Филтрирање података

Када имамо пуно података, некада је потребно да издвојимо оне који нам могу бити од користи. томе служи филтрирање података. Ево примера како да из низа који смо малопре формирали издвојимо све бројеве који су изнад просека:

```python
просек_низа=avg(низ)
натпросечни=[x for x in низ if x >= просек_низа]
```

Сада можемо видети који су то натпросечни елементи низа:

```python
натпросечни
```

Као што смо и знали, имамо два елемента кој су натпросечни. На овај начин можемо филтрирати елементе по било ком другом критеријуму и тако олакшати посао када радимо са великим количинама података.

### 13.4 Приказивање података

Да бисмо у неким ситуацијама видели нешто што се не види из низа и табеле, потребно је да визуализујемо податке. Посматрајмо пример групе дечака који су шутирали тројке. Низовима су представљена њихова имена и број убачених тројки из 25 покушаја.

```python
дечаци=["Милан","Стефан","Павле","Ненад","Жељко","Борис","Дарко"]
број_погодака=[10,7,11,13,17,15,21]
```

Помоћу хистограма можемо лепше видети како изгледа ова листа, па ћемо прво нацртати најједноставнији хистограм.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,7))
plt.bar(дечаци,број_погодака)
plt.title("Број убачених тројки")
plt.show()
plt.close()
```

Сада је згодно да са датог дијаграма видимо ко је шутирао изнад просека, па ћемо зато додати и једну линију која означава просек. Да бисмо то урадили неопходно је да га израчунамо и да га прикажемо у свакој тачки дијаграма.

```python
просек=avg(број_погодака)
```

Сада ћемо приказато комбиновани график.

```python
plt.figure(figsize=(7,7))
plt.bar(дечаци,број_погодака,label="број_погодака")
plt.plot(дечаци, [просек]*len(дечаци), color="red",label="просек")
plt.title("Број убачених тројки")
plt.legend()
plt.show()
plt.close()
```

Посматрајмо сада низ који смо раније користили да прикажемо велико одступање медијане од просека. Хајде да пробамо да га прикажемо помоћу хистограма.

```python
n = len(низ)
plt.bar(range(n),низ)
plt.show()
plt.close()
```

Овакав график нам не значи ништа јер за већину података не можемо ништа да очитамо. Зато ћемо применити неколико трикова и график ће добити смисао.

```python
n = len(низ)
plt.figure(figsize=(15,8))
plt.bar(range(n), низ, color="b")
plt.plot([0, n-1], [просек_низа, просек_низа], color="r")
plt.ylim(0,60)
plt.show()
plt.close()
```
